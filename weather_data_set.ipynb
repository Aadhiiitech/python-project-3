```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/historical_weather_data.csv")
```

```python
data.head()
```

```python
data.isnull().sum()

data = data.dropna()
```

```python
data.info()

data.describe()
```

```python
import matplotlib.pyplot as plt

plt.plot(data["Temperature"])
plt.title("Temperature Trend Over Time")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.show()
```

```python
plt.hist(data["Temperature"], bins=20)
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(data["Temperature"], data["Humidity"])
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.show()
```

```python
import matplotlib.pyplot as plt

weather_condition_counts = data["WeatherCondition"].value_counts()

plt.pie(weather_condition_counts, labels=weather_condition_counts.index, autopct="%1.1f%%")
plt.title("Weather Condition Distribution")
plt.show()
```
